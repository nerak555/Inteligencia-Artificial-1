DATASET de CICO

```python
# PASO 1: Nuestro Dataset de CICO
# X = Cantidad de estudiantes que ingresaron
# Y = Cantidad de estudiantes que se titularon

X = [120, 150, 135, 180, 200, 210, 190, 170, 220, 250, 240, 260, 280, 160, 145, 300, 290, 275, 310, 320, 330, 230, 205, 175, 195]
Y = [55, 68, 62, 85, 90, 98, 88, 75, 105, 115, 110, 120, 135, 70, 65, 145, 140, 125, 150, 155, 160, 100, 95, 80, 87]

print("Total de datos cargados:", len(X))
```

Entrenando el Algoritmo

```python
# PASO 2: Variables para guardar a los "ganadores"
mejor_b = 0
mejor_w = 0
menor_error = 9999999 # Empezamos con un error gigante para que el primer intento lo supere

print("Iniciando el entrenamiento de fuerza bruta...")

# PASO 3: Los bucles (Estructuras repetitivas)
# Evaluaremos 'b' desde -50 hasta 50
for b in range(-50, 50):

    # Evaluaremos 'w' (Como Python 'for' no usa decimales directo, probamos del 0 al 100 y dividimos)
    for w_int in range(0, 100):
        w = w_int / 100 # Esto probará pesos desde 0.00 hasta 0.99

        error_total_del_intento = 0

        # Recorremos nuestros 25 datos históricos
        for i in range(len(X)):
            x_actual = X[i] # Los inscritos reales
            y_real = Y[i]   # Los titulados reales

            # ¡LA FÓRMULA MATEMÁTICA! (Nuestra predicción)
            y_prediccion = b + (w * x_actual)

            # Calculamos la diferencia entre la realidad y la predicción (Error Absoluto)
            error = abs(y_real - y_prediccion)
            error_total_del_intento = error_total_del_intento + error

        # Estructura de control decisivo (El condicional)
        # Si esta combinación de b y w logró un error menor, los guardamos como los mejores
        if error_total_del_intento < menor_error:
            menor_error = error_total_del_intento
            mejor_b = b
            mejor_w = w

print("¡Entrenamiento terminado en segundos!")
print("--- RESULTADOS ---")
print("El mejor Sesgo (b) encontrado es:", mejor_b)
print("El mejor Peso (w) encontrado es:", mejor_w)
print("Margen de error total acumulado:", menor_error)
```

Las 10 Inferencias (El final de la tarea)

```python
# PASO 4: Inferencias (Prediciendo el futuro)
print("\n--- PREDICCIONES PARA EL FUTURO DE CICO ---")

# Inventamos 10 cantidades nuevas de estudiantes ingresados (Nuevos valores de X)
nuevos_ingresos = [350, 380, 400, 420, 450, 480, 500, 550, 600, 1000]

for x_nuevo in nuevos_ingresos:
    # Usamos nuestra fórmula entrenada: y = b + wx
    prediccion_titulados = mejor_b + (mejor_w * x_nuevo)

    # Redondeamos el resultado porque no podemos tener "medios" estudiantes
    titulados_redondeado = round(prediccion_titulados)

    print(f"Si ingresan {x_nuevo} estudiantes, se espera que se titulen aprox: {titulados_redondeado}")
```
